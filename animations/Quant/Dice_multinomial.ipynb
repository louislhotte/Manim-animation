# I - Dice multinomial

```python
class Intro(Scene):
    def construct(self):
        intro_group = self.introduction("Exactly 6 [Quant Interview]", 
                                            "Python | Brainteaser | Probabilities")
        self.play(FadeOut(intro_group))

    def introduction(self, title1, title2):
        header = Tex(title1)
        header.set_width(8)
        from_pos = [header.get_left()[0] - 1, header.get_bottom()[1] - 0.5, 0]
        to_pos = [header.get_right()[0] + 1, header.get_bottom()[1] - 0.5, 0]
        line = Line(from_pos, to_pos)
        writer = Tex("Created by Ptolémé")
        writer_pos = [(line.get_left()[0] + line.get_right()[0]) / 2, line.get_bottom()[1] - 1, 0]
        writer.move_to(writer_pos)
        
        self.play(Write(header), Write(line))
        self.wait(2)
        self.play(Transform(header, Tex(title2)))
        self.play(Write(writer))
        self.wait(1)
        
        return VGroup(header, writer, line)
    
    def Outro(self, text):
        header = Tex(text)

class Scene1_Countdown(Scene):
    def construct(self):
        # Display the question
        question = Tex(
            r"If I roll a dice 36 times, what is the probability \\ that each number appears exactly 6 times?",
            font_size=36
        )
        
        # Counter label and number
        counter_label = Tex("Answer in ", font_size=36)
        counter_number = Integer(20, font_size=36)
        counter_group = VGroup(counter_label, counter_number).arrange(RIGHT)
        
        # Arrange question and counter vertically
        full_group = VGroup(question, counter_group).arrange(DOWN, buff=1)
        self.add(full_group)
        
        # Animate countdown from 20 to 0 over 20 seconds
        time_left = ValueTracker(20)
        counter_number.add_updater(lambda d: d.set_value(int(time_left.get_value())))
        self.play(
            time_left.animate.set_value(0),
            run_time=20,
            rate_func=linear
        )
        counter_number.remove_updater()

class Scene2_Formula(Scene):
    def construct(self):
        # General probability formula
        formula1 = Tex(
            r"P = \frac{\text{Number of favorable outcomes}}{\text{Total number of outcomes}}",
            font_size=40
        )
        self.play(Write(formula1))
        self.wait(2)
        
        # Total outcomes
        total = Tex(r"\text{Total outcomes} = 6^{36}", font_size=40)
        total.next_to(formula1, DOWN, buff=1)
        self.play(Write(total))
        self.wait(2)
        
        # Favorable outcomes (multinomial coefficient)
        favorable = Tex(r"\text{Favorable outcomes} = \frac{36!}{(6!)^6}", font_size=40)
        favorable.next_to(total, DOWN, buff=1)
        self.play(Write(favorable))
        self.wait(2)
        
        # Final probability formula
        probability = Tex(r"P = \frac{36!}{(6!)^6 \cdot 6^{36}}", font_size=40)
        probability.next_to(favorable, DOWN, buff=1)
        self.play(Write(probability))
        self.wait(3)

class Scene3_Explanation(Scene):
    def construct(self):
        # Total outcomes
        total = Tex(r"6^{36} \text{ total sequences}", font_size=36)
        self.play(Write(total))
        self.wait(2)
        
        # Step-by-step favorable outcomes
        step1 = Tex(
            r"\text{Choose positions for each number:}",
            font_size=36
        )
        step2 = Tex(
            r"C(36,6) \cdot C(30,6) \cdot C(24,6) \cdot C(18,6) \cdot C(12,6) \cdot C(6,6)",
            font_size=30
        )
        step1.next_to(total, DOWN, buff=1)
        step2.next_to(step1, DOWN, buff=0.5)
        self.play(Write(step1), Write(step2))
        self.wait(3)
        
        # Simplify to multinomial coefficient
        step3 = Tex(
            r"= \frac{36!}{6! \cdot 30!} \cdot \frac{30!}{6! \cdot 24!} \cdots \frac{6!}{6! \cdot 0!}",
            font_size=28
        )
        step4 = Tex(r"= \frac{36!}{6! \cdot 6! \cdot 6! \cdot 6! \cdot 6! \cdot 6!}", font_size=36)
        step5 = Tex(r"= \frac{36!}{(6!)^6}", font_size=36)
        
        step3.next_to(step2, DOWN, buff=0.5)
        step4.next_to(step3, DOWN, buff=0.5)
        step5.next_to(step4, DOWN, buff=0.5)
        
        self.play(Write(step3))
        self.wait(2)
        self.play(Write(step4))
        self.wait(2)
        self.play(Write(step5))
        self.wait(2)
        
        # Final probability
        final_formula = Tex(r"P = \frac{36!}{(6!)^6 \cdot 6^{36}}", font_size=40)
        final_formula.next_to(step5, DOWN, buff=1)
        self.play(Write(final_formula))
        self.wait(2)
        
        # Numerical approximation
        approx = Tex(r"\approx 0.000258", font_size=40, color=YELLOW)
        approx.next_to(final_formula, DOWN, buff=0.5)
        self.play(Write(approx))
        self.wait(3)

class Scene4_Thanks(Scene):
    def construct(self):
        thanks = Tex(r"Thanks for watching!", font_size=48, color=BLUE)
        self.play(Write(thanks))
        self.wait(2)
```
