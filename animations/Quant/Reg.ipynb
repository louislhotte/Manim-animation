<b>[Bloomberg - Easy]</b> Say you draw a circle and choose two chords at random. What is the probability that those chords will intersect?

    <strong>Solution:</strong> Imagine you choose four points at random on the circle’s circumference. You then form two chords by pairing these points. The chords will intersect if the endpoints are arranged in an alternating order around the circle (for instance, A, C, B, D rather than A, B, C, D). A combinatorial argument shows that out of the three equally likely ways to pair up four points into two chords, only one results in intersecting chords. Hence, the probability is 1/3.

```python
%%manim -ql -v WARNING --disable_caching InterpretLinearRegression


class InterpretLinearRegression(Scene):
    def construct(self):
        # 1) Generate and fit a linear regression on scatter data
        np.random.seed(0)
        xs = np.linspace(0, 4, 12)
        ys = 2 * xs + 1 + np.random.normal(0, 0.5, len(xs))

        axes = Axes(
            x_range=[0, 4, 1], y_range=[0, 10, 2],
            x_length=6, y_length=4,
            axis_config={"include_tip": False}
        ).to_edge(DOWN)

        points = VGroup(*[
            Dot(axes.c2p(x, y), radius=0.05, color=BLUE)
            for x, y in zip(xs, ys)
        ])

        # Fit model
        slope, intercept = np.polyfit(xs, ys, 1)
        y_pred = slope * xs + intercept
        residuals = ys - y_pred
        sse = np.sum(residuals**2)
        n = len(xs)
        sxx = np.sum((xs - np.mean(xs))**2)
        se_slope = np.sqrt(sse/(n-2)/sxx)
        t_slope = slope / se_slope

        line_graph = axes.plot(lambda x: slope * x + intercept, color=RED)

        # Animate regression
        self.play(Create(axes), run_time=1)
        self.play(FadeIn(points), run_time=1.5)
        self.play(Create(line_graph), run_time=2)

        # Display fitted formula and t-statistic
        formula = MathTex(
            rf"y = b_1 x + b_0 \\\\ b_1 = {slope:.2f},\quad b_0 = {intercept:.2f}",
            font_size=28
        ).to_edge(UP)
        t_formula = MathTex(
            rf"t_{{b_1}} = \frac{{b_1}}{{\mathrm{{SE}}(b_1)}} = {t_slope:.2f}",
            font_size=28
        ).next_to(formula, DOWN)

        self.play(Write(formula), Write(t_formula), run_time=2)
        self.wait(1.5)
        self.play(*[FadeOut(mob) for mob in self.mobjects], run_time=1)

        # 2) Title and subtitle for t-statistics overview
        title = Text("Interpreting Linear Regression", font_size=48)
        subtitle = Text("Feature significance via t-statistics", font_size=24).next_to(title, DOWN)
        self.play(Write(title), run_time=1.5)
        self.play(FadeIn(subtitle), run_time=1)
        self.wait(1)
        self.play(FadeOut(title), FadeOut(subtitle), run_time=1)

        # 3) Bar chart of feature t-values
        features = ["Feature 1", "Feature 2"]
        t_values = [2.3, 5.1]
        bar_chart = BarChart(
            values      = t_values,
            bar_names   = features,
            y_range     = [0, 6, 1],
            x_length    = 6,
            y_length    = 4,
            bar_colors  = [BLUE, GREEN],
        ).scale(0.8).to_edge(DOWN)

        bar_labels = VGroup(*[
            Text(f"t={t:.1f}", font_size=24)
            .next_to(bar, UP, buff=0.1)
            for bar, t in zip(bar_chart.bars, t_values)
        ])
        self.play(Create(bar_chart), Write(bar_labels), run_time=2)
        self.wait(1)

        max_idx = int(np.argmax(t_values))
        box = SurroundingRectangle(bar_chart.bars[max_idx], buff=0.1, color=YELLOW)
        note = Text(
            f"{features[max_idx]} is most significant (t={t_values[max_idx]:.1f})",
            font_size=24
        ).next_to(bar_chart, UP)
        self.play(Create(box), Write(note), run_time=2)
        self.wait(1)

        # 4) Explanation of why use t-statistics
        why_title = Text("Why use t-statistics?", font_size=36).to_edge(UP)
        points_list = BulletedList(
            "Accounts for feature scale",
            "Normalizes by estimation uncertainty",
            "Provides unitless, comparable scores",
            font_size=24
        ).next_to(why_title, DOWN, aligned_edge=LEFT)

        self.play(
            FadeOut(bar_chart), FadeOut(bar_labels), FadeOut(box), FadeOut(note),
            run_time=1
        )
        self.play(Write(why_title), Write(points_list), run_time=3)
        self.wait(2)
        self.play(*[FadeOut(mob) for mob in self.mobjects], run_time=1.5)
```
